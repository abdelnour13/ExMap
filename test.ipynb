```python
from functools import wraps

class AddOption(type):

    def __new__(cls, name, bases, dct):

        cls = super().__new__(cls, name, bases, dct)

        orig_options = cls.options

        @wraps(orig_options.__func__)
        def new_options(cls):
            return orig_options() + [44]
                
        cls.options = classmethod(new_options)

        return cls

class ConcreteStage(AddOption):

    @classmethod
    def options(cls) -> list[int]:
        return [5, 5]
    
class ChildStage(ConcreteStage):

    @classmethod
    def options(cls) -> list[int]:
        return [4, 4]
```
